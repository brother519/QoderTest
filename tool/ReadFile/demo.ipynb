```python
import pandas as pd
import numpy as np

data = pd.DataFrame({'name': ['Alice', 'Bob', 'Charlie'], 'score': [85, 92, 78]})
print(data)
```

## Data Analysis

Basic data analysis with pandas.

```python
mean_score = data['score'].mean()
print(f'Mean: {mean_score:.1f}')
```

```python
high = data[data['score'] > 80]
print(high)
```
